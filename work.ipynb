# Outlier Detection

```python
Q1=data["Total_price"].quantile(0.25)
Q3=data["Total_price"].quantile(0.75)
IQR=Q3-Q1
lower,upper=Q1-1.5*IQR,Q3+1.5*IQR
outliers=data[(data["Total_price"]<lower)|(data["Total_price"]>upper)]
outliers.shape
```

```python
data=data[(data["Total_price"]>=lower) & (data ["Total_price"] <=upper)]
```

```python
data['order_date'] = pd. to_datetime(data['order_date'])
monthly_sales = data.groupby (
    data['order_date'].dt.month
)['Total_price'].sum()
monthly_sales.plot(kind='line', figsize=(10,4), title='Monthly Sales Trend')
```

```python
region_sales = data.groupby('region') ['revenue'].sum().sort_values(ascending=False)
region_sales
```

```python
region_sales=data.groupby("region")["revenue"].mean()
print(region_sales)
plt.bar(region_sales.index,region_sales.values,color="green",edgecolor="k")
plt.show()
```

```python
category_discount=data.groupby("product_category")["discount"].mean()
plt.bar(category_discount.index,category_discount.values)
plt.xlabel("Product Category")
plt.ylabel("average discount")
plt.title("average discount by product category")
plt.show()
```
